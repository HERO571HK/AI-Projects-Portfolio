```python
# Iris datasetini sklearn üzerinden yüklüyoruz

import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df.head()
```

```python
# Dataset hakkında bilgi alıyoruz

df.info()
```

```python
# Sayısal istatistikleri inceliyoruz

df.describe()
```

```python
# Eksik veri kontrolü yapıyoruz

df.isnull().sum()
```

```python
# Veriyi görselleştiriyoruz

import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df)

plt.show()
```

```python
# K-Means modelini kuruyoruz

from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

kmeans.fit(df)
```

```python
# Cluster sonuçlarını oluşturuyoruz

df["Cluster"] = kmeans.labels_

df.head()
```

```python
# Cluster görselleştirmesi yapıyoruz

plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df["petal length (cm)"],
    y=df["petal width (cm)"],
    hue=df["Cluster"],
    palette="deep"
)

plt.title("K-Means Iris Clustering")

plt.show()
```

```python
# Cluster sayılarını inceliyoruz

df["Cluster"].value_counts()
```
