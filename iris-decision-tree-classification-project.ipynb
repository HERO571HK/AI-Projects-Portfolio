```python
# Iris datasetini yüklüyoruz

import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target

df.head()
```

# Veri Setinin İncelenmesi

Bu bölümde veri setinin yapısı, eksik veri durumu ve istatistiksel özellikleri incelenmektedir.

```python
# Eksik veri kontrolü yapıyoruz

df.isnull().sum()
```

# Bağımlı ve Bağımsız Değişkenlerin Ayrılması

Bu bölümde target sütunu ayrılarak modelleme için veri hazırlanmaktadır.

```python
# X ve y oluşturuyoruz

X = df.drop("target", axis=1)

y = df["target"]
```

# Train Test Split İşlemi

Bu bölümde veri eğitim ve test verisi olarak ayrılmaktadır.

```python
# Train test split yapıyoruz

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

# Decision Tree Modelinin Kurulması

Bu bölümde Decision Tree algoritması kullanılmaktadır.

```python
# Decision Tree modelini kuruyoruz

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

dt_pred = dt_model.predict(X_test)

dt_acc = accuracy_score(y_test, dt_pred)

print("Decision Tree Accuracy:", dt_acc)
```

# Confusion Matrix ve Classification Report

Bu bölümde model performansı detaylı olarak incelenmektedir.

```python
# Confusion matrix ve classification report alıyoruz

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

print(confusion_matrix(y_test, dt_pred))

print(classification_report(y_test, dt_pred))
```
