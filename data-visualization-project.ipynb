```python
# Titanic datasetini KaggleHub ile indiriyoruz

import kagglehub

dataset_path = kagglehub.dataset_download(
    "heptapod/titanic"
)

print(dataset_path)
```

```python
# Dataset dosyalarını kontrol ediyoruz

import os

print(os.listdir(dataset_path))
```

```python
# Dataseti yüklüyoruz

import pandas as pd

df = pd.read_csv(dataset_path + "/train_and_test2.csv")

df.head()
```

```python
# Dataset bilgilerini kontrol ediyoruz

df.info()
```

```python
# Sayısal istatistikleri inceliyoruz

df.describe()
```

```python
# Gerekli kütüphaneleri import ediyoruz

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
# Yaş dağılımını görselleştiriyoruz

plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=30)

plt.title("Age Distribution")

plt.show()
```

```python
# Yaş ve ücret ilişkisini görselleştiriyoruz

plt.figure(figsize=(8,5))

sns.scatterplot(x=df["Age"], y=df["Fare"])

plt.title("Age vs Fare")

plt.show()
```

```python
# Korelasyon haritasını oluşturuyoruz

plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
