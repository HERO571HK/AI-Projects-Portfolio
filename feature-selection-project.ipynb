# Proje Yol Haritası

1. Dataset yükleme
2. X ve y oluşturma
3. Feature Selection uygulama
4. En önemli değişkenleri bulma
5. Sonuç değerlendirme

```python
# Gerekli kütüphaneler

import pandas as pd

from sklearn.datasets import load_iris
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

# Dataset

iris = load_iris()

X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

y = iris.target

# Feature Selection

selector = SelectKBest(
    score_func=f_classif,
    k=2
)

X_new = selector.fit_transform(X,y)

scores = pd.DataFrame({
    "Özellik":X.columns,
    "Skor":selector.scores_
})

scores = scores.sort_values(
    by="Skor",
    ascending=False
)

print(scores)
```
