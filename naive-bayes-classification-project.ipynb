```python
# Iris datasetini yüklüyoruz

import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target

df.head()
```

# Veri Setinin İncelenmesi

Bu bölümde veri setinin yapısı ve istatistiksel özellikleri incelenmektedir.

```python
# Veri yapısını inceliyoruz

df.info()
```

```python
# İstatistiksel özet alıyoruz

df.describe()
```

```python
# Eksik veri kontrolü yapıyoruz

df.isnull().sum()
```

# X ve y Oluşturma

Bu bölümde bağımlı ve bağımsız değişkenler ayrılmaktadır.

```python
# X ve y oluşturuyoruz

X = df.drop("target", axis=1)

y = df["target"]
```

# Train Test Split İşlemi

Bu bölümde veri eğitim ve test verisi olarak ayrılmaktadır.

```python
# Train test split yapıyoruz

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

# Naive Bayes Modelinin Kurulması

Bu bölümde Gaussian Naive Bayes algoritması kullanılmaktadır.

```python
# Naive Bayes modelini kuruyoruz

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

nb_model = GaussianNB()

nb_model.fit(X_train, y_train)

nb_pred = nb_model.predict(X_test)

nb_acc = accuracy_score(y_test, nb_pred)

print("Naive Bayes Accuracy:", nb_acc)
```

# Confusion Matrix ve Classification Report

Bu bölümde model performansı detaylı olarak incelenmektedir.

```python
# Confusion matrix ve classification report alıyoruz

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

print(confusion_matrix(y_test, nb_pred))

print(classification_report(y_test, nb_pred))
```
