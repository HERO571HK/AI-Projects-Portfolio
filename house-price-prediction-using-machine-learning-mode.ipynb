```python
# Streamlit uygulama dosyasını oluşturuyoruz

streamlit_code = '''

import streamlit as st
import joblib
import pandas as pd

model = joblib.load("house_price_model.pkl")

st.title("House Price Prediction App")

overallqual = st.slider("Overall Quality", 1, 10, 5)
grlivarea = st.number_input("Ground Living Area", 500, 5000, 1500)
garagecars = st.slider("Garage Cars", 0, 5, 2)
totalbsmtsf = st.number_input("Total Basement SF", 0, 3000, 800)

input_df = pd.DataFrame({
    "OverallQual":[overallqual],
    "GrLivArea":[grlivarea],
    "GarageCars":[garagecars],
    "TotalBsmtSF":[totalbsmtsf]
})

prediction = model.predict(input_df)

st.subheader("Predicted House Price")

st.write(f"${prediction[0]:,.0f}")

'''

with open("app.py", "w") as f:
    f.write(streamlit_code)

print("app.py oluşturuldu.")
```

## Streamlit Uygulaması İçin Basit Model Oluşturma

Bu aşamada web uygulamasında kullanıcıdan alınacak 4 temel konut özelliği ile çalışan daha sade bir Random Forest modeli oluşturulacaktır.

```python
# Streamlit için gerekli veriyi yeniden okuyup sade modeli kaydediyoruz

import pandas as pd
import joblib
from sklearn.ensemble import RandomForestRegressor

DATA_PATH = "/kaggle/input/competitions/house-prices-advanced-regression-techniques"

train_df = pd.read_csv(DATA_PATH + "/train.csv")

simple_features = ["OverallQual", "GrLivArea", "GarageCars", "TotalBsmtSF"]

X_simple = train_df[simple_features]
y = train_df["SalePrice"]

simple_model = RandomForestRegressor(random_state=42)

simple_model.fit(X_simple, y)

joblib.dump(simple_model, "house_price_model.pkl")

print("Streamlit için sade model kaydedildi.")
```

```python
# requirements.txt oluşturuyoruz

with open("requirements.txt", "w") as f:
    f.write("streamlit\n")
    f.write("scikit-learn\n")
    f.write("pandas\n")
    f.write("joblib\n")

print("requirements.txt oluşturuldu.")
```
