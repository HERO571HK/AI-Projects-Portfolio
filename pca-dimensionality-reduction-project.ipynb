# PCA Dimensionality Reduction Project

## Projenin Amacı

Bu projede Iris veri seti kullanılarak PCA (Principal Component Analysis) yöntemi uygulanacaktır.

Amaç, çok boyutlu veriyi daha az boyuta indirerek görselleştirmek ve verinin temel yapısını daha anlaşılır hale getirmektir.

## Proje Yol Haritası

2. Veri setinin incelenmesi
3. Bağımlı ve bağımsız değişkenlerin ayrılması
4. Verilerin ölçeklendirilmesi
5. PCA uygulanması
6. Verinin 2 boyuta indirgenmesi
7. PCA sonucunun görselleştirilmesi
8. Açıklanan varyans oranının incelenmesi
9. Proje sonuçlarının yorumlanması

```python
# Iris datasetini yüklüyoruz

import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target

df.head()
```

# Veri Setinin İncelenmesi

Bu bölümde veri setinin yapısı, istatistiksel özellikleri ve eksik veri durumu incelenmektedir.

```python
# Veri yapısını inceliyoruz

df.info()
```

```python
# Sayısal istatistikleri inceliyoruz

df.describe()
```

```python
# Eksik veri kontrolü yapıyoruz

df.isnull().sum()
```

# Bağımlı ve Bağımsız Değişkenlerin Ayrılması

Bu bölümde target sütunu ayrılmıştır.

```python
# X ve y oluşturuyoruz

X = df.drop("target", axis=1)

y = df["target"]
```

# Verilerin Ölçeklendirilmesi

Bu bölümde PCA öncesinde veriler standardize edilmektedir.

```python
# Verileri ölçeklendiriyoruz

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

# PCA Uygulanması

Bu bölümde veri 2 boyuta indirgenmektedir.

```python
# PCA modelini kuruyoruz

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
# PCA dataframe oluşturuyoruz

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

pca_df["target"] = y

pca_df.head()
```

# PCA Sonuçlarının Görselleştirilmesi

Bu bölümde PCA ile elde edilen 2 boyutlu veri görselleştirilmektedir.

```python
# PCA görselleştirmesi yapıyoruz

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.scatterplot(
    x=pca_df["PC1"],
    y=pca_df["PC2"],
    hue=pca_df["target"],
    palette="deep"
)

plt.title("PCA Iris Visualization")

plt.show()
```

# Açıklanan Varyans Oranı

Bu bölümde PCA bileşenlerinin veriyi açıklama oranları incelenmektedir.

```python
# Açıklanan varyans oranını inceliyoruz

print(pca.explained_variance_ratio_)
```
